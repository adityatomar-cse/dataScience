Question1. Create a Python datetime object representing 15 February 2026 at 10:30 AM. Display the year,
month, day, hour and minute separately.

```python
from datetime import datetime

dt = datetime(2026, 2, 15, 10, 30)

print("Year:", dt.year)
print("Month:", dt.month)
print("Day:", dt.day)
print("Hour:", dt.hour)
print("Minute:", dt.minute)
```

Question2. Create two datetime objects representing 10:00 AM and 12:30 PM on 15 February 2026. Calculate
the difference between them in hours and minutes.

```python
from datetime import datetime

t1 = datetime(2026, 2, 15, 10, 0)
t2 = datetime(2026, 2, 15, 12, 30)

diff = t2 - t1

print("Difference in hours:", diff.seconds // 3600)
print("Difference in minutes:", diff.seconds // 60)
```

Question3. Convert the string "20/02/2026" into a datetime object using strptime(). Display the result in
YYYY-MM-DD format using strftime().

```python
from datetime import datetime

date = datetime.strptime("20/02/2026", "%d/%m/%Y")

print(date.strftime("%Y-%m-%d"))
```

Question4. Convert the following list of date strings into Pandas datetime values using pd.to_datetime():
dates = ["2026-02-01", "2026-02-05", "2026-02-10"]

```python
import pandas as pd

dates = ["2026-02-01", "2026-02-05", "2026-02-10"]

dates = pd.to_datetime(dates)

print(dates)
```

Q5. Convert the following values into datetime objects. Handle invalid and missing dates using
errors="coerce" and identify the missing values.
dates = ["2026-02-01", "invalid", "2026-02-05", None]

```python
import pandas as pd

dates = ["2026-02-01", "invalid", "2026-02-05", None]

dates = pd.to_datetime(dates, errors="coerce")

print(dates)
print("Missing values:")
print(dates.isna())
```

Question11. Generate a DatetimeIndex containing 10 consecutive daily dates, starting from 1 March 2026,
using pd.date_range().

```python
import pandas as pd

dates = pd.date_range(
    start="2026-03-01",
    periods=10,
    freq="D"
)

print(dates)
```
