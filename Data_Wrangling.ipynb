#Hierarchical Indexing (MultiIndex)

```python
import pandas as pd

# create the student data
data = {
    'Department': ['CSE', 'CSE', 'DS', 'DS'],
    'Year': [1, 2, 1, 2],
    'Enrolled_Students': [100, 120, 80, 90]
}


 #create the DataFrame
df = pd.DataFrame(data)
#Department and year to index
result = df.set_index(['Department', 'Year'])

print(result)
```

#Merge students and marks

```python
import pandas as pd

students = pd.DataFrame({
    'Student_ID': [101, 102, 103],
    'Name': ['Aman', 'Ravi', 'Neha']
})

marks = pd.DataFrame({
    'Student_ID': [101, 102, 104],
    'Marks': [85, 90, 75]
})

print("Inner Join:")
print(pd.merge(students, marks, on='Student_ID', how='inner'))

print("Left Join:")
print(pd.merge(students, marks, on='Student_ID', how='left'))

print("Outer Join:")
print(pd.merge(students, marks, on='Student_ID', how='outer'))
```

#Concatenate monthly sales

```python
import pandas as pd

jan = pd.DataFrame({
    'Month': ['Jan', 'Jan'],
    'Product': ['Pen', 'Book'],
    'Sales': [100, 200]
})

feb = pd.DataFrame({
    'Month': ['Feb', 'Feb'],
    'Product': ['Pen', 'Book'],
    'Sales': [150, 250]
})

result = pd.concat([jan, feb], ignore_index=True)

print(result)
```

#Long format to Wide format using pivot()


```python
import pandas as pd

data = {
    'Product': ['Pen', 'Pen', 'Book', 'Book'],
    'Quarter': ['Q1', 'Q2', 'Q1', 'Q2'],
    'Revenue': [100, 150, 200, 250]
}

df = pd.DataFrame(data)

result = df.pivot(
    index='Product',
    columns='Quarter',
    values='Revenue'
)

print(result)
```

#pivot_table() – Calculate total revenue

```python
import pandas as pd

data = {
    'Product': ['Pen', 'Pen', 'Pen', 'Book', 'Book'],
    'Quarter': ['Q1', 'Q1', 'Q2', 'Q1', 'Q2'],
    'Revenue': [100, 150, 200, 300, 400]
}

df = pd.DataFrame(data)

result = df.pivot_table(
    index='Product',
    columns='Quarter',
    values='Revenue',
    aggfunc='sum'
)

print(result)
```

#Wide format to Long format using melt()

```python
import pandas as pd

data = {
    'Student': ['Aman', 'Ravi', 'Neha'],
    'Test_1': [80, 75, 90],
    'Test_2': [85, 80, 95],
    'Test_3': [90, 85, 88]
}

df = pd.DataFrame(data)

result = df.melt(
    id_vars='Student',
    var_name='Test',
    value_name='Marks'
)

print(result)
```

#Stack and Unstack

```python
import pandas as pd

data = {
    'Semester_1': [76, 82],
    'Semester_2': [81, 85]
}

df = pd.DataFrame(data, index=['CSE', 'DS'])

print("Stack:")
stacked = df.stack()
print(stacked)

print("Unstack:")
print(stacked.unstack())
```

#Merge attendance and marks

```python
import pandas as pd

profiles = pd.DataFrame({
    'Student_ID': [201, 202, 203, 204],
    'Name': ['Aman', 'Ravi', 'Neha', 'Sara'],
    'Department': ['CSE', 'DS', 'CSE', 'DS']
})

attendance = pd.DataFrame({
    'Student_ID': [201, 202, 203, 204],
    'Attendance': [92, 78, 85, 96]
})

marks = pd.DataFrame({
    'Student_ID': [201, 202, 203, 204],
    'Final_Marks': [88, 73, 81, 94]
})

df = profiles.merge(attendance, on='Student_ID')
df = df.merge(marks, on='Student_ID')

print("Combined Data:")
print(df)

print("Department Average:")
print(df.groupby('Department')['Final_Marks'].mean())
```
